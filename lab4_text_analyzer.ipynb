### <span style="color:crimson">Lists: an ordered collection</span>
A **list** holds several values, in order, inside square brackets. Each item has a **position**, called its **index**, and Python counts from **0**.

```
scores = [72, 88, 95, 61, 79]
index:     0   1   2   3   4
negative: -5  -4  -3  -2  -1
```

Negative indexes count from the end, so `scores[-1]` is always the **last** item, however long the list is.

````python
scores = [72, 88, 95, 61, 79]
print(scores[0])      # 72  - the first item
print(scores[2])      # 95
print(scores[-1])     # 79  - the last item
print(len(scores))    # 5   - how many items
print(88 in scores)   # True - is it in the list?
````

### <span style="color:crimson">Slicing: a piece of a list</span>
`items[start:stop]` makes a **new list** from `start` up to, **but not including**, `stop` — the same rule as `range()`. Leave out `start` to begin at the front; leave out `stop` to go to the end.

````python
scores = [72, 88, 95, 61, 79]
print(scores[1:3])    # [88, 95]        - positions 1 and 2, not 3
print(scores[:2])     # [72, 88]        - the first two
print(scores[2:])     # [95, 61, 79]    - from position 2 to the end
print(scores[1:-1])   # [88, 95, 61]    - everything except the two ends
print(scores[:3] == [72, 88, 95])   # True
````

### <span style="color:crimson">Changing a list</span>
Lists are **mutable**: you can change them after they are made.

| You write | What happens |
|---|---|
| `items[0] = 100` | replaces the first item |
| `items.append(5)` | adds 5 to the end |
| `items.sort()` | puts the list in order, **in place** |
| `items.index(88)` | the position of the first 88 |

<span style="color:crimson">__A trap worth knowing:__</span> a function that receives a list receives **the same list**, not a copy. If the function changes it, the caller sees the change. That is sometimes what you want (Exercise 1's `replace_first`) and sometimes a nasty surprise. A slice, by contrast, is always a **new** list.

````python
def add_bonus(items):
    items.append(100)          # changes the caller's list

grades = [70, 80]
add_bonus(grades)
print(grades)                  # [70, 80, 100] - the original list changed

top_two = grades[:2]           # a NEW list
top_two[0] = 0
print(grades, top_two)         # [70, 80, 100] [0, 80] - grades is untouched
````

### <span style="color:crimson">Strings are sequences too</span>
A string is a sequence of characters, so indexing and slicing work exactly as they do for lists: `"Python"[0]` is `"P"` and `"Python"[-3:]` is `"hon"`.

One big difference: strings are **immutable**. You cannot change a character in place; every string method gives you back a **new** string.

| Method | Example | Result |
|---|---|---|
| `lower()` / `upper()` | `"Hi There".lower()` | `"hi there"` |
| `strip(chars)` | `"--hi!--".strip("-!")` | `"hi"` — removes from **both ends only** |
| `split()` | `"a  b\nc".split()` | `["a", "b", "c"]` — splits on any spaces or line breaks |
| `split(".")` | `"x.y.z".split(".")` | `["x", "y", "z"]` |
| `join(list)` | `"-".join(["a", "b"])` | `"a-b"` |
| `replace(old, new)` | `"a!b!".replace("!", ".")` | `"a.b."` |
| `startswith(s)` | `"horse".startswith("h")` | `True` |

````python
line = "  Programs become INTERESTING when they handle collections!  "
print(line.strip())                      # spaces gone from both ends
print(line.lower().split())              # a list of lowercase words
print("hello!!".strip("!"))              # hello
print("don't".strip("'"))                # don't  - the inside apostrophe stays
print(" ".join(["one", "two", "three"])) # one two three
````

### <span style="color:crimson">Lining text up with f-strings</span>

### <span style="color:crimson">The three iteration patterns</span>
Nearly every loop that walks through a collection is doing one of three jobs.

<img src="module4/iteration_patterns.svg" alt="Accumulate turns many values into one result; search stops at the first match; filter keeps only the items that qualify, in order" width="760">

**Accumulate** — start with an empty result, and fold each item into it.
```python
total = 0
for x in numbers:
    total = total + x        # after the loop, total holds the answer
```
**Search** — look at items until one matches, then **stop**. The "not found" answer comes **after** the loop.
```python
for x in numbers:
    if x < 0:
        return x             # found it: stop looking
return None                  # only reached if nothing matched
```
**Filter** — build a **new** list of just the items that qualify, keeping their order.
```python
result = []
for x in numbers:
    if x > 3:
        result.append(x)
```

<span style="color:crimson">__The most common search bug__</span> is putting the "not found" answer **inside** the loop with an `else`:
```python
for word in words:
    if word.startswith("h"):
        return word
    else:
        return ""        # WRONG: gives up after looking at only ONE word
```
Trace it with `["cat", "horse"]`: it checks `"cat"`, fails, and returns `""` without ever reaching `"horse"`.

### <span style="color:crimson">Counting with two lists</span>
To count how often each word appears, you need to remember **two** things per word: the word and its count. You have not met *dictionaries* yet (that is Module 5), and you do not need them: two lists that stay **side by side** do the job.

Here is the technique on a different problem, counting votes:

````python
votes = ["Bo", "Ana", "Bo", "Chidi", "Ana", "Bo"]

names = []     # names[i] ...
tallies = []   # ... has tallies[i] votes

for vote in votes:
    if vote in names:                       # SEARCH: have we seen it before?
        position = names.index(vote)
        tallies[position] = tallies[position] + 1   # ACCUMULATE its count
    else:
        names.append(vote)                  # first time: start at 1
        tallies.append(1)

print(names)     # ['Bo', 'Ana', 'Chidi']
print(tallies)   # [3, 2, 1]
````

Trace the first four votes to see the two lists grow together:

| vote | seen before? | `names` after | `tallies` after |
|---|---|---|---|
| Bo | no | `['Bo']` | `[1]` |
| Ana | no | `['Bo', 'Ana']` | `[1, 1]` |
| Bo | yes, position 0 | `['Bo', 'Ana']` | `[2, 1]` |
| Chidi | no | `['Bo', 'Ana', 'Chidi']` | `[2, 1, 1]` |

### <span style="color:crimson">Sorting pairs, and breaking ties</span>
Put each name and its tally together as a **pair**, a small list `[name, count]`, and you can sort the pairs.

* `pairs.sort()` sorts pairs by their **first** item, so alphabetically by name.
* `pairs.sort(key=by_count, reverse=True)` sorts by whatever the function `by_count` returns — here, the count — **highest first**.

Python's sort is **stable**: when two pairs have the same count, the second sort leaves them in the order the first sort put them. So *sort alphabetically first, then by count*, and ties come out in alphabetical order. Two short lines, and no tie-breaking code at all.

````python
def by_count(pair):
    return pair[1]           # the count part of [name, count]

pairs = [["Bo", 3], ["Dee", 2], ["Ana", 2], ["Chidi", 1]]
pairs.sort()                              # alphabetical: Ana, Bo, Chidi, Dee
pairs.sort(key=by_count, reverse=True)    # by count, ties stay alphabetical
print(pairs)        # [['Bo', 3], ['Ana', 2], ['Dee', 2], ['Chidi', 1]]
print(pairs[:2])    # the top two
````
